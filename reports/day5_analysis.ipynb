## 1. Load the cleaned data


```python
from pathlib import Path
import pandas as pd
from IPython.display import display
base = Path.cwd()
if not (base / 'data' / 'clean_data.csv').exists():
    base = base.parent
df = pd.read_csv(base / 'data' / 'clean_data.csv', parse_dates=['start_date'])
display(df.head())
```

## 3. Numeric summary, groups, and correlations


```python
numeric = ['salary', 'performance_score', 'experience_years']
display(df[numeric].agg(['mean', 'median', 'min', 'max', 'std']))
display(df.groupby('department').agg(employees=('id','count'), avg_salary=('salary','mean')))
display(df[numeric].corr())
```

## 5. Department budget merge + pivot table


```python
budgets = pd.read_csv(base / 'data' / 'department_budgets.csv')
merged = pd.merge(df, budgets, on='department', how='left')
display(merged[['name','department','annual_budget_inr']].head())
display(pd.pivot_table(df,index='department',columns='city',values='salary',aggfunc='mean',fill_value=0))
```
